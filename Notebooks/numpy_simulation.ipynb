```python
import numpy as np
```

```python
dice_rolls = np.random.randint(1, 7, 10000)
print("\n FIRST 20 DICE ROLLS")
print(dice_rolls[:20])
```

```python
expected_value = np.mean(dice_rolls)
print("\nExpected Value:")
print(expected_value)
```

```python
variance = np.var(dice_rolls)
print("\n VARIANCE")
print(variance)
```

```python
unique,counts = np.unique(dice_rolls,return_counts=True)
print(unique,counts)
```

```python
for numbers , count in zip(unique,counts):
    print(f"{numbers} comes {count} times")

print("\n Probability :")

prob = counts / 10000

for number, probability in zip(unique, prob):
   print(f"P({number}) = {probability}")
```
