```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(7)
print(np.random.randint(1, 10, 5))
```

```python
data=pd.DataFrame({
    "Machine" :np.random.choice(["A","B"],size=100),
    "Defective": np.random.choice(["Yes","No"],size=100,p=[0.2,0.8])})
data.head()
```

```python
# calculate probability of defective products

prob_defective = (data["Defective"] == "Yes").mean()

print("Probability of defective product:", prob_defective)
```

```python
p_a = (data["Machine"] == "A").mean()
print("Probability of Machine A:",p_a)
```

```python
p_b = (data["Machine"]  == "B").mean()
print("Probability of Machine B:",p_b)       
```

```python
# Conditional Probability
machine_a = data[data["Machine"]=="A"]
p_def_a = (machine_a["Defective"]=="Yes").mean()
print("Probability of Defective Machine A :",p_def_a)

machine_b = data[data["Machine"]=="B"]
p_def_b = (machine_b["Defective"]=="Yes").mean()
print("Probability of Defective Machine B :",p_def_b)
```

```python
# Bayes’ Theorem
p_d = prob_defective
p_a_given_d = (p_def_a * p_a) / p_d
p_b_given_d = (p_def_b * p_b) / p_d

print("P(Machine A | Defective):", p_a_given_d)
print("P(Machine B | Defective):", p_b_given_d)
```

```python
data["Defective_Report"] = data["Defective"].map({"Yes": 1, "No": 0})
data.head()
```

```python
data["Defective_Report"].value_counts().plot(kind="bar")

plt.title("Defect Distribution (0 = No, 1 = Yes)")
plt.xlabel("Defective (0 = No, 1 = Yes)")
plt.ylabel("Count")

plt.show()
```

# Interpretation :

1.Calculating the average of A and B gives the probability of each machine being used in production.
2.Finding the conditional probability of A with defective products shows the defect rate specifically for Machine A.
3.Finding the conditional probability of B with defective products shows the defect rate specifically for Machine B.
4.Bayes’ Theorem helps identify the most likely source (Machine A or B) when a defective product is observed.
Distribution visualization shows how defective and non-defective products are spread across the dataset.
