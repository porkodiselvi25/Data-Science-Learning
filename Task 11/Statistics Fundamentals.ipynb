```python
outliers = pd.DataFrame({
    "Marks": [300, 400, 500]
})
```

```python
new_marks = pd.concat(
    [df["Marks"], outliers["Marks"]],
    ignore_index=True
)
```

```python
print(new_marks)
```

```python
print("Mean:", new_marks.mean())
print("Median:", new_marks.median())
```

```python
plt.hist(new_marks, bins=10)

plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```

```python
sns.boxplot(x=new_marks)

plt.title("Outliers in Marks")
plt.show()
```

# Dataset after adding outliers
1. Mean increased a lot after adding outliers.
2. Median changed only slightly.
3. Variance and standard deviation increased.
4. Data became positively skewed.
5. Outliers affected mean more than median.
6. Study hours and marks showed a positive correlation.
