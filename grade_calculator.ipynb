```python
# task 3
# grade_calculator.py

print("-------CALCULATING THE GRADE-------\n")

def get_score(score):
    if score > 90:
        return "A GRADE"
    elif score > 80:
        return "B GRADE"
    elif score > 70:
        return "C GRADE"
    elif score > 50:
        return "D GRADE"
    else:
        return "FAIL"

while True:

    name = input("ENTER YOUR NAME : ")

    if name == "stop":
        break

    score = float(input("ENTER YOUR MARKS : "))

    grade = get_score(score)

    print("NAME :", name)
    print("SCORE :", score)
    print("GRADE : \n", grade)

    print("TYPE 'stop' to stop \n")
```
