```python
import pandas as pd

customer_data = pd.read_csv(r"C:\Users\admin\PORKODI CLG\Data-Science-Learning\Customer Data\data\Raw customer.csv")
print(customer_data)
```

```python
#clear null data
print(customer_data.isnull().sum())
customer_data.dropna()
```

```python
print(customer_data.duplicated().sum())
```

```python
customer_data = customer_data.drop_duplicates()
print(customer_data)
```

```python
customer_data["Age"]=customer_data["Age"].fillna(customer_data["Age"].mean())
print(customer_data["Age"])
```

```python
customer_data["City"] = customer_data["City"].fillna("unknown")
print(customer_data["City"])
```

```python
#remove the extra space in city
customer_data["City"] = customer_data["City"].str.strip()
print(customer_data["City"])
```

```python
customer_data["City"].str.title()
```

```python
customer_data["Gender"].str.title()
```

```python
customer_data["Salary"] = pd.to_numeric(customer_data["Salary"], errors="coerce")
print(customer_data["Salary"])
```

```python
customer_data["Salary"] = customer_data["Salary"].fillna(customer_data["Salary"].mean())
print(customer_data["Salary"])
```

```python
customer_data = customer_data[(customer_data["Age"] > 0) & (customer_data["Age"] <= 100)]
print(customer_data)
```

```python
customer_data.to_csv("customerdata.csv")
```
