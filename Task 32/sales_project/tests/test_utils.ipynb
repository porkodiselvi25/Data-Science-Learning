```python
import logging
```

```python
def setup_logger():
    logging.basicConfig(
        level=logging.INFO,
        format="%(asctime)s - %(levelname)s - %(message)s")

    return logging.getLogger()
```

```python
def test_logger():
    logger = setup_logger()
    
    assert logger is not None
    
    print("Logger test passed")


test_logger()
```
