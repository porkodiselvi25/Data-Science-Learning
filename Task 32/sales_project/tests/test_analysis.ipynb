```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Name": ["A", "B", "C"],
    "Sales": [100, 200, 300]})

print(df)
```

```python
def test_get_shape():
    assert df.shape == (3, 2)

test_get_shape()

print("Shape test passed")
```

```python
def test_get_columns():
    assert list(df.columns) == ["Name", "Sales"]

test_get_columns()

print("Columns test passed")
```

```python
def test_missing_values():
    assert df.isnull().sum().sum() == 0

test_missing_values()

print("Missing value test passed")
```
