```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
np.random.seed(42)

group_A =np.random.normal(70,10,50)
group_B =np.random.normal(75,10,50)
```

```python
t_stat,p_value = stats.ttest_ind(group_A,group_B)
print("t_stat:",t_stat)
print("p_value:",p_value)
```

```python
alpha = 0.5

if p_value < alpha:
    print("Reject Null Hypothesis ")
    print("Conclusion: Significant difference between groups")

else:
    print("Fail to Reject Null Hypothesis")
    print("Conclusion: No significant difference between groups")
```

```python
print("Mean A:", group_A.mean())
print("Mean B:", group_B.mean())
print("Difference:", abs(group_A.mean() - group_B.mean()))
```

```python
observed = np.array([
    [30, 20],
    [40, 10]
])

chi2, p_chi, dof, expected = stats.chi2_contingency(observed)

print("Chi-square:", chi2)
print("p-value:", p_chi)
```

```python
print("\nFINAL RESULT")

if p_value < alpha:
    print("t-test: Significant difference found")
else:
    print("t-test: No significant difference")

if p_chi < alpha:
    print("chi-square: Significant relationship found")
else:
    print("chi-square: No significant relationship found")
```

# Decision Memo
Objective:

- Compare two groups to check if there is a meaningful difference.

Method:
- Independent t-test is used to compare the mean of two groups
- Chi-square test is used to check relationship between categorical variables
- Significance level  = 0.05

Results:
- t-test p-value checks difference in averages  
- Chi-square p-value checks relationship between categories  

Conclusion:
- If p < 0.05 → statistically significant difference  
- If p ≥ 0.05 → no strong evidence of difference  

Practical Meaning:
- Even if results are significant, we check whether the difference is large enough to matter in real-world use.

Final Decision"
- Reject or accept null hypothesis based on p-value
- Decide whether to adopt new method or not
