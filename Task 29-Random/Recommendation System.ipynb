```python
import pandas as pd
import numpy as np

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python

movies = pd.DataFrame({

    "Movie":["Avengers","Iron Man","Captain America","Batman","Superman","Spiderman","Doctor Strange","Thor"],
    "Genre":["Action Superhero Sci-Fi","Action Superhero Technology","Action Superhero War","Action DC Crime","Action DC Alien","Action Superhero Marvel","Magic Superhero Fantasy","Action Mythology Marvel"],     
    "Rating":[4.9,4.7,4.6,4.5,4.2,4.8,4.4,4.3]
})
print(movies)
```

```python
popular = movies.sort_values(by="Rating", ascending=False)
print(popular[["Movie","Rating"]])
```

```python
tfidf = TfidfVectorizer()
genre_matrix = tfidf.fit_transform(movies["Genre"])
print(genre_matrix)
```

```python
similarity = cosine_similarity(genre_matrix)

print(similarity)
```

```python
def recommend(movie_name):
    index = movies[movies["Movie"] == movie_name].index[0]
    scores = list(enumerate(similarity[index]))
    scores = sorted(scores,key=lambda x: x[1],reverse=True)
    scores = scores[1:6]
    print("Recommendations for", movie_name)

    for i in scores:
        print(movies.iloc[i[0]]["Movie"])
        
recommend("Avengers")
recommend("Batman")
```
