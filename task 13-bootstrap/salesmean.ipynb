```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
sales=np.random.normal(loc=5000,scale=800,size=100)
data=pd.DataFrame({"Sales": sales})
data.head()
```

```python
plt.hist(data["Sales"],bins=20)
plt.title("SALES DISTRIBUTION")
plt.xlabel("sales")
plt.ylabel("Frequency")
plt.show()
```

```python
def bootstrap_means(data, n_bootstrap=1000):
    means = []

    for i in range(n_bootstrap):
        sample = np.random.choice(data, size=len(data), replace=True)
        means.append(np.mean(sample))

    return np.array(means)
bootstrap_distribution = bootstrap_means(data["Sales"])
print(bootstrap_distribution[:10])
```

```python
plt.hist(bootstrap_distribution,bins=20)
plt.title("Bootstrap Distribution")
plt.xlabel("Means sales")
plt.ylabel("frequency")
plt.show()
```

```python
lower=np.percentile(bootstrap_distribution,2.5)
upper=np.percentile(bootstrap_distribution,97.5)
print("95% Confidence Interval:")
print("LOWER =",lower)
print("UPPER =",upper)
```

```python
standard_error = np.std(bootstrap_distribution)

print("Standard Error:", standard_error)
```

```python
margin_of_error = (upper - lower) / 2

print("Margin of Error:", margin_of_error)
```
