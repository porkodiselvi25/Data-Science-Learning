```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("loan_data.csv")

print("Original Dataset:")
print(df.head())
```

```python
print("\nDataset Info")
print(df.info())
```

```python
df["income_to_loan_ratio"] = df["income"] / (df["loan_amount"] + 1)

df["loan_per_income"] = df["loan_amount"] / (df["income"] + 1)

print("\nNUMERIC FEATURES")
print(
    df[
        [
            "income",
            "loan_amount",
            "income_to_loan_ratio",
            "loan_per_income"
        ]
    ].head()
)
```

```python
encoder = LabelEncoder()

df["employment_type_encoded"] = encoder.fit_transform(
    df["employment_type"]
)

print("\nCATEGORICAL FEATURE")
print(
    df[
        [
            "employment_type",
            "employment_type_encoded"
        ]
    ].head()
)
```

```python
df["application_date"] = pd.to_datetime(df["application_date"])

df["application_year"] = df["application_date"].dt.year

df["application_month"] = df["application_date"].dt.month

df["application_day"] = df["application_date"].dt.day

df["application_weekday"] = df["application_date"].dt.day_name()

print("\nDATETIME FEATURES")
print(
    df[
        [
            "application_date",
            "application_year",
            "application_month",
            "application_day",
            "application_weekday"
        ]
    ].head()
)
```

```python
df["remarks_length"] = df["remarks"].astype(str).apply(len)

df["word_count"] = df["remarks"].astype(str).apply(
    lambda x: len(x.split())
)

print("\nTEXT FEATURES")
print(
    df[
        [
            "remarks",
            "remarks_length",
            "word_count"
        ]
    ].head()
)
```

```python
df["age_group"] = pd.cut(
    df["age"],
    bins=[18, 25, 35, 50, 100],
    labels=["Young", "Adult", "Middle", "Senior"]
)

print("\nBINNING FEATURE")
print(
    df[
        [
            "age",
            "age_group"
        ]
    ].head()
)
```

```python
df["age_income_interaction"] = (
    df["age"] * df["income"]
)

print("\nINTERACTION FEATURE")
print(
    df[
        [
            "age",
            "income",
            "age_income_interaction"
        ]
    ].head()
)
```

```python
print("\nNUMERIC FEATURES")
display(
    df[
        [
            "income",
            "loan_amount",
            "income_to_loan_ratio",
            "loan_per_income"
        ]
    ].head()
)
```

```python
print("\nLEAKAGE AWARENESS")

print("""
Only application-time information is used.

Safe Features:
- age
- income
- loan_amount
- employment_type
- application_date

Excluded Leakage Features:
- future payment status
- future default status
- recovery amount
""")
```

```python
data_dictionary = pd.DataFrame({
    "Feature": [
        "age",
        "income",
        "loan_amount",
        "employment_type",
        "application_date",
        "Loan_Default",
        "income_to_loan_ratio",
        "loan_per_income",
        "employment_type_encoded",
        "application_year",
        "application_month",
        "application_day",
        "application_weekday",
        "remarks_length",
        "word_count",
        "age_group",
        "age_income_interaction"
    ],

    "Type": [
        "Numeric",
        "Numeric",
        "Numeric",
        "Categorical",
        "Datetime",
        "Target",
        "Engineered Numeric",
        "Engineered Numeric",
        "Encoded Categorical",
        "Datetime",
        "Datetime",
        "Datetime",
        "Datetime",
        "Text",
        "Text",
        "Binning",
        "Interaction"
    ]
})

print("\nDATA DICTIONARY")
print(data_dictionary)
```

```python
df.to_csv(
    "feature_engineered_loan_data.csv",
    index=False
)

data_dictionary.to_csv(
    "data_dictionary.csv",
    index=False
)

print("\nFiles Saved Successfully!")
```
