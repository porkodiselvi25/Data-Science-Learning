```python
#dict
Student = {
    "Name" : "Laksh",
    "course" : "Python"
}

for key,value in Student.items():
    print(key,":",value)
```

```python
#nested dictionaries
student = {
    "student1" : {
        "Name" : "Laksh",
        "course" : "M.sc IT"
    },
    "student2" : {
        "Name" : "Porkodi",
        "course" : "M.sc IT"
    }
}

print(student)

print(student["student1"]["Name"])
print(student["student2"]["Name"])

for key,value in student.items():
    print(key,":",value)
```
