```python
import numpy as np
```

```python
data=np.array([[1, 2, 3],
    [2, 4, 6],
    [3, 1, 0],
    [5, 2, 1],
    [0, 1, 1]])
print("dataset:\n",data)
```

```python
v1 = data[0]
v2 = data[1]

similarity = cosine_similarity(v1, v2)
print("\nCosine Similarity between v1 and v2:", similarity)
```

```python
def cosine_similarity(a,b):
    dot_product=np.dot(a,b)
    norm_a = np.linalg.norm(a)
    norm_b = np.linalg.norm(b)
    return dot_product/(norm_a*norm_b)
```

```python
def nearest_neighbor(query_vector, dataset):
    similarities = []

    for vector in dataset:
        sim = cosine_similarity(query_vector, vector)
        similarities.append(sim)

    best_index = np.argmax(similarities)

    return best_index, dataset[best_index], similarities
```

```python
query = np.array([1, 2, 2])
```

```python
index, nearest_vector, sims = nearest_neighbor(query, data)
```

```python
print("\nQuery Vector:", query)
print("Nearest Vector:", nearest_vector)
print("Index of Nearest Vector:", index)
```

```python
print("\nSimilarity Scores:")
for i, sim in enumerate(sims):
    print(f"Vector {i}: {sim}")
```
