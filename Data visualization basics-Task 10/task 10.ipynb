```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df=sns.load_dataset("titanic")
df.head()
```

```python
df.info()
```

```python
#Create Histogram
'''Histogram is best for:
      numerical data
      distribution
      frequency analysis
  Used here to see passenger age distribution.'''


plt.figure(figsize=(4,2))
plt.hist(df["age"].dropna())
plt.title("Age Distribution")
plt.show()
```

```python
#Create Pie Chart
'''Pie chart is best for:
      percentage
      part-to-whole relationship

  Used here to show:
      male percentage
      female percentage'''

Gender = df["sex"].value_counts()
plt.pie(Gender,labels=Gender.index,autopct="%1.1f%%")
plt.title("Gender Distribution")
plt.show()
```

```python
#Create Bar Chart
'''Bar chart is used for:
    category comparison
    count comparison

  Used here to compare:
    survived
    not survived'''

sns.countplot(data=df , x ="survived",hue="survived")
plt.title("Survived")
plt.show()
```

```python
#Create Scatter Plot
'''Scatter plot is used to find:
      relationships
      correlation
      trends
      
   Used here to compare:age,fare'''

plt.scatter(df["age"],df["fare"])
plt.title("Age vs Fare")
plt.xlabel("age")
plt.ylabel("fare")
plt.show()
```

```python
# Create Line Chart
'''Line chart is best for:
        trends
        ordered data
Used here because passenger classes are ordered:
1 → 2 → 3'''


avgr_fare=df.groupby("pclass")["fare"].mean()
plt.plot(avgr_fare.index ,avgr_fare.values)
plt.title("pclass Fare")
plt.xlabel("pclass")
plt.ylabel("Fare")
plt.show()
```

```python
#Create Box Plot
'''Box plot is useful for:
        finding outliers
        understanding spread
        comparing distributions

Used here to compare fare spread across classes.'''

    
sns.boxplot(x="pclass",y="fare",data=df)
plt.title("pclass Fare")
plt.show()
```

```python
#Create Heatmap
'''Heatmap is best for:
        correlation analysis
        comparing many numerical columns together

Used here to identify relationships between variables.'''

sns.heatmap(df.corr(numeric_only=True),annot=True)
plt.title("correlation Heatmap")
plt.show()
```

```python
#Create Violin Plot
'''Violin plot is useful for:
            distribution
            density
            spread analysis

Used here to compare age distribution between genders.'''

sns.violinplot(x="sex",y="age",data=df)
plt.title("Age Distriution")
plt.show()
```
